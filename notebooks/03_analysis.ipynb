```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Load the processed weather data
data_path = Path('../data/processed/weather_data.parquet')
df = pd.read_parquet(data_path)

# Display basic information about the dataset
print("Dataset Info:")
print("-" * 40)
print(df.info())
print("\nSample of the data:")
print(df.head())

# Basic statistical summary
print("\nStatistical Summary:")
print("-" * 40)
print(df.describe())

# Create visualizations
plt.style.use('seaborn')
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# Temperature distribution
sns.histplot(data=df, x='temperature', bins=30, ax=axes[0,0])
axes[0,0].set_title('Temperature Distribution')
axes[0,0].set_xlabel('Temperature (°C)')

# Temperature trends over time
sns.lineplot(data=df, x='date', y='temperature', ax=axes[0,1])
axes[0,1].set_title('Temperature Trends')
axes[0,1].set_xlabel('Date')
axes[0,1].set_ylabel('Temperature (°C)')

# Humidity vs Temperature scatter plot
sns.scatterplot(data=df, x='temperature', y='humidity', ax=axes[1,0])
axes[1,0].set_title('Humidity vs Temperature')
axes[1,0].set_xlabel('Temperature (°C)')
axes[1,0].set_ylabel('Humidity (%)')

# Monthly average temperatures
monthly_temps = df.groupby(df['date'].dt.month)['temperature'].mean()
monthly_temps.plot(kind='bar', ax=axes[1,1])
axes[1,1].set_title('Average Temperature by Month')
axes[1,1].set_xlabel('Month')
axes[1,1].set_ylabel('Average Temperature (°C)')

plt.tight_layout()
plt.show()

# Correlation analysis
correlation_matrix = df.corr()
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Matrix of Weather Variables')
plt.show()
```
